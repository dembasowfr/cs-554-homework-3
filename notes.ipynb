### How Bias Nodes Work
Bias nodes are typically added to each layer of a neural network. They are initialized with a value (often zero) and are updated during the training process through backpropagation. The mathematical representation can be illustrated as follows:

$
    y = f(W * x + b)
$

Where:
y is the output,

f is the activation function,

W is the weight matrix,

x is the input vector,

b is the bias term.

This equation shows how the bias term b shifts the output of the activation function, allowing for more complex mappings from inputs to outputs.

## 🔹 Theoretical Explanation

A **bias node** is analogous to the **intercept term** in a linear equation (like $y = mx + b $). Just as the intercept allows the line to shift up or down, **bias allows the activation function of a neuron to be shifted**, enabling the model to better fit the data.

Without a bias term, all activation functions (like sigmoid, tanh, ReLU) would be constrained to pass through the origin, limiting the representational power of the network.

### Why is bias important?
- It lets the model **fit data that is not centered around the origin**.
- It allows the network to learn functions **even when all inputs are zero**.
- Biases increase the **flexibility** of the network by shifting the activation threshold.

## 🔹 Mathematical Explanation

Consider a simple neuron in a single-layer perceptron. The output of a neuron $z $ before applying the activation function is:

$
z = \sum_{i=1}^n w_i x_i + b
$

Where:
- $ x_i $ are the input features,
- $w_i $ are the weights,
- $b $ is the **bias term**,
- $z $ is the weighted sum,
- and the final output is $a = \phi(z) $, where $\phi $ is the activation function.

### Example:

For a single neuron with inputs $x_1, x_2 $, weights $w_1, w_2 $, and bias $b $:

$
z = w_1 x_1 + w_2 x_2 + b \\
a = \phi(z)
$

If $\phi $ is the sigmoid function:
$
a = \frac{1}{1 + e^{-z}} = \frac{1}{1 + e^{-(w_1 x_1 + w_2 x_2 + b)}}
$

### Vector Form:
For a layer with input vector $\mathbf{x} \in \mathbb{R}^n $, weight vector $\mathbf{w} \in \mathbb{R}^n $, and scalar bias $b $:

$
z = \mathbf{w}^\top \mathbf{x} + b
$

For multiple neurons (say in a dense layer with $m $ outputs), we use a **bias vector** $\mathbf{b} \in \mathbb{R}^m $:

$
\mathbf{z} = \mathbf{W} \mathbf{x} + \mathbf{b}
$

- $\mathbf{W} \in \mathbb{R}^{m \times n} $: weight matrix
- $\mathbf{x} \in \mathbb{R}^{n} $: input vector
- $\mathbf{b} \in \mathbb{R}^{m} $: bias vector
- $\mathbf{z} \in \mathbb{R}^{m} $: pre-activation outputs

---

### Without Bias

$
z = \mathbf{w}^\top \mathbf{x}
$

The output would always be zero when all inputs are zero, which is a constraint that reduces model flexibility.

### 🔍 **Mathematical Explanation**

This is a **1-hidden-layer neural network** with:
- Input \( x \in \mathbb{R} \)
- Hidden layer with 2 units using **tanh** activation
- Linear output for regression

#### **Forward Pass**

1. Add bias to input: \( \tilde{x} = [x, 1] \)
2. Hidden layer:  
\[
z_h = \tanh(w_h^T \tilde{x}) \quad \text{for } h = 1, 2
\]
3. Add bias to hidden output: \( \tilde{z} = [z_1, z_2, 1] \)
4. Output:  
\[
y = v^T \tilde{z}
\]

#### **Loss Function**
\[
E = \frac{1}{2} (r - y)^2
\]

#### **Backward Pass (Gradients)**

Let \( \delta = r - y \)

- For output weights:
\[
\Delta v_j = \eta \cdot \delta \cdot z_j
\]

- For hidden weights:
\[
\Delta w_{hj} = \eta \cdot \delta \cdot v_h \cdot (1 - z_h^2) \cdot x_j
\]
*(Because derivative of tanh is \( 1 - z_h^2 \))*
